**Step-4: Frequency table creation**

```python
import os
new_folder_name= 'bank_dataset_frequency_tables'
os.makedirs(new_folder_name)
```

```python
cwd = os.getcwd()
new_dir= os.path.join(cwd, new_folder_name)
############################################################
for i in cat[1:]:
    bank_df[i].value_counts()
    keys= bank_df[i].value_counts().keys()
    values= bank_df[i].value_counts().values
    df= pd.DataFrame(zip(keys, values),
                    columns= ['Count','Number of Applicants'])
##############################################################
    file_name= f'{i}.csv'
    new_path= os.path.join(new_dir, file_name)
    df.to_csv(new_path, index= False)
```

```python
types= bank_df.dtypes.to_dict()
cat= [i for i,j in types.items() if j== 'object']
```

```python
#Making directory
import os
folder_name= "bank_bar_charts"
try:
    os.makedirs(folder_name)
except Exception as e:
    print(e)

cwd= os.getcwd()
new_dir= os.path.join(cwd, folder_name)

for i in cat[1:]:
    keys= bank_df[i].value_counts().keys()
    values= bank_df[i].value_counts().values
    file_name= f'{i}_bar.jpg'
    path= os.path.join(new_dir, file_name)
    plt.bar(keys, values)
    #plt.pie(keys, values)
    plt.savefig(path)
```
