```python
def Emperical_rule(k,n):
    wage_mean= visa_df['prevailing_wage'].mean()
    wage_std= visa_df['prevailing_wage'].std()
    lb= wage_mean-k*wage_std
    ub= wage_mean+k*wage_std
    con1= visa_df['prevailing_wage'] > lb
    con2= visa_df['prevailing_wage'] < ub
    con= con1 & con2
    print(len(visa_df[con])== n*len(visa_df)/100)

```

**Outlier imputation**

- method1: 
    - less than lb outlier impute with lb value
    - greater than ub outlier impute with ub value

- method2: 
    - Every outlier replace with median

```python
l=[]
for i in wage_data:
    if i< lb:
        l.append(lb)
    elif i> ub:
        l.append(ub)
    else:
        l.append(i)

visa_df['wage_data_new']=l
```
