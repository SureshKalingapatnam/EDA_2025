```python
#Variance
l1= [1, 100, 2, 98, 3, 95, 4, 5, 120, 77]
mean= np.mean(l1)
arr1= np.array(l1)
s1= arr1- mean
s2= np.square(s1)
s3= np.sum(s2)
s3/len(l1)

np.sum(np.square(arr1-mean))/len(l1)
```

```python
#Standard deviation
l1= [1, 100, 2, 98, 3, 95, 4, 5, 120, 77]
mean= np.mean(l1)
arr1= np.array(l1)
s1= arr1- mean
s2= np.square(s1)
s3= np.sum(s2)
s3/len(l1)

np.sqrt(np.sum(np.square(arr1-mean))/len(l1))
```

```python
l1= [10,20,30,40]
l2= [20,32,40,50]
arr1= np.array(l1)
arr2= np.array(l2)
mean_x= np.mean(l1)
mean_y= np.mean(l2)
np.sum((arr1- mean_x)*(arr2-mean_y))/len(l1)
```
