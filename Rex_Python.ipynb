**re.findall**
- using a \d+ pattern
- \d indicates digit match between 0 to 9
- plus indicates one or more occurances of digit

```python
str1= " the price is 100 and the tax is 20"
for i in str1.split():
    if i.isdigit():
        print(i)
```

```python
pattern= r"\d+"
str1= " the price is 100 and the tax is 20"
result= re.findall(pattern, str1)
result
```

```python
pattern= r"\d+"
str1= "Order number 1234 shipped on 28th December 2024"
result= re.findall(pattern, str1)
print(result)
```

```python
# ^ : Used for start of the string
# [a-zA-Z0-9_.+-]: pattern if any letter in email if have its
                   not only once charcater , many character
                    for that we need to add a + sign at last
# till now the pattern is ^[a-zA-Z0-9_.+-]
# till now the patterns is ^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+
#\.
# till now the pattern is ^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+\.

#.com
# [a-zA-Z]{2,}
# till now the pattern is ^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+\.[a-zA-Z]{2,}

# $: end of string

```

```python
pattern= r"^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9]+\.[a-zA-Z]{2,}$"
email= "Suresh.Kalingapatnam@gmail.com"
bool(re.match(pattern, email))
```

*Matching a sequqnce of 10 numbers**

```python
import re
pattern= r"\d{3}" # exactly 3 digits match
str1= "my mobile number is 1234567890"
re.findall(pattern, str1)
```

```python
import re
pattern= r"\d{3,}" #minimum 3 digits match
str1= "my mobile number is 1234567890 456 1245 12"
re.findall(pattern, str1)
```

```python
import re
pattern= r"\d{3}|[0-9]" # exactly 3 digits match
str1= "my mobile number is 1234567890"
re.findall(pattern, str1)
```

```python
1234567890
\d+ : 1234567890
\d{3}: 123,456,789
\d{3}| [0-9] 123, 456,789, 0
```

```python
import re
pattern= r"\d{15}" # exactly 3 digits match
str1= "my mobile number is 123456789012345"
re.findall(pattern, str1)
```

```python
import re
pattern= r"\d{3}-\d{3}-\d{2}" # exactly 3 digits match
str1= "my mobile number is 123-456-89"
re.findall(pattern, str1)
```

```python
import re
pattern= r'\(\d{3}\) \d{3}-\d{3}' # exactly 3 digits match
str1= "my mobile number is (123) 456-7890"
re.findall(pattern, str1)
```

```python
str1= 'my number is 123-456-7890, (123) 456-789'
pattern= r'\d{3}-\d{3}-\d{3} | \(\d{3}\) \d{3}-\d{3}'
re.findall(pattern, str1)
```

```python
str1='my number is 123-456-7890,(123) 456-7890,123.456.7890'
pattern= r'\(?\d{3,}\)?[-.\s]\d{3,}[-.\s]\d{3,}'
re.findall(pattern, str1)
```
